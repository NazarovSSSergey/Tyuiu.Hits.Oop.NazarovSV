### Описание задачи:
Создать базовый класс OrderLine в C#, который будет представлять информацию о строке заказа, содержащей детали одного товара в заказе. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
1. СтандартнаяСтрока (StandardLine): Должна содержать дополнительные атрибуты, такие как Количество единиц (Units). Метод CalculateTotal() должен быть переопределен для учета количества единиц при расчете общей стоимости.
2. СпециальнаяСтрока (SpecialLine): Должна содержать дополнительные атрибуты, такие как Скидка (Discount). Метод UpdatePrice() должен быть переопределен для применения скидки к цене товара.
3. БесплатнаяСтрока (FreeLine) (если требуется третий класс): Должна содержать дополнительные атрибуты, такие как Предварительный платеж (Prepayment). Метод CalculateTotal() должен быть переопределен для учета предварительного плата при расчете общей стоимости.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

OrderLine line1 = new StandardLine(1, "Товар А", 100m, 3);
OrderLine line2 = new SpecialLine(2, "Товар Б", 200m, 10);
OrderLine line3 = new FreeLine(3, "Товар В", 150m, 50m);

line1.GetProductDetails();
Console.WriteLine("Итого: " + line1.CalculateTotal());

line2.UpdatePrice(200m);
line2.GetProductDetails();
Console.WriteLine("Итого: " + line2.CalculateTotal());

line3.GetProductDetails();
Console.WriteLine("Итого: " + line3.CalculateTotal());

Console.WriteLine("Общая сумма заказа: " + (line1.CalculateTotal() + line2.CalculateTotal() + line3.CalculateTotal()));
Console.WriteLine();

StandardLine line4 = new StandardLine(4, "Товар Г", 500m, 2);
line4.Category = "Продукты";
line4.Weight = 5;
line4.MarkAsPaid();
line4.ShowStatus();
Console.WriteLine("Общий вес: " + line4.GetTotalWeight());
Console.WriteLine("Дорогой товар: " + line4.IsExpensive());
Console.WriteLine("Большая партия: " + line4.IsBulk());
line4.ShowPackaging();
Console.WriteLine();

SpecialLine line5 = new SpecialLine(5, "Товар Д", 1200m, 15);
line5.PromoCode = "SALE15";
line5.ShowPromo();
Console.WriteLine("Дорогой товар: " + line5.IsExpensive());
Console.WriteLine("Большая скидка: " + line5.IsBigDiscount());
Console.WriteLine("Сумма скидки: " + line5.GetDiscountAmount());
Console.WriteLine();

FreeLine line6 = new FreeLine(6, "Товар Е", 150m, 150m);
line6.Reason = "Подарок";
line6.ShowReason();
Console.WriteLine("Полностью оплачено: " + line6.IsFullyPrepaid());
Console.WriteLine("Остаток к оплате: " + line6.GetDebt());
Console.WriteLine();

WholesaleLine line7 = new WholesaleLine(7, "Товар Ж", 100m, 10, 20);
line7.Warehouse = "Склад 2";
line7.GetProductDetails();
Console.WriteLine("Итого: " + line7.CalculateTotal());
Console.WriteLine("Экономия: " + line7.GetSavedMoney());
Console.WriteLine("Оптовая покупка: " + line7.IsWholesale());
line7.ShowWarehouse();
Console.WriteLine("Общий вес: " + line7.GetTotalWeight());
line7.MarkAsPaid();
Console.WriteLine();

DeliveryLine line8 = new DeliveryLine(8, "Товар З", 300m, "ул. Ленина, 5", 14);
line8.Courier = "Иван";
line8.DeliveryCost = 50m;
line8.GetProductDetails();
line8.Deliver();
line8.ReturnItem();
Console.WriteLine("Итого: " + line8.CalculateTotal());
Console.WriteLine("Итого с доставкой: " + line8.GetTotalWithDelivery());
Console.WriteLine();

Console.WriteLine("--- Перегрузка методов ---");
line1.UpdatePrice(120m, "Новая цена поставщика");
line1.ShowStatus("Статус первой строки:");
Console.WriteLine("Итого с тремя доп. единицами: " + line4.CalculateTotal(3));
Console.WriteLine("Сумма скидки на 3 штуки: " + line5.GetDiscountAmount(3));
Console.WriteLine("Остаток с доп. платежом: " + ((FreeLine)line3).GetDebt(20m));
line8.Deliver("ул. Мира, 10");
Console.WriteLine("Итого с доп. стоимостью доставки: " + line8.GetTotalWithDelivery(25m));
Console.WriteLine();

Console.WriteLine("--- Generic классы ---");
OrderCollection<OrderLine> allLines = new OrderCollection<OrderLine>();
allLines.Add(line1);
allLines.Add(line2);
allLines.Add(line3);
allLines.DisplayLines();
Console.WriteLine("Сумма всех строк: " + allLines.GetTotalSum());
Console.WriteLine();

OrderCollection<StandardLine> standardLines = new OrderCollection<StandardLine>();
standardLines.Add(line4);
standardLines.Add(line7);
standardLines.DisplayLines();
Console.WriteLine("Сумма стандартных строк: " + standardLines.GetTotalSum());
Console.WriteLine();

OrderCollection<DeliveryLine> deliveryLines = new OrderCollection<DeliveryLine>();
deliveryLines.Add(line8);
deliveryLines.DisplayLines();
Console.WriteLine();

allLines.Remove(line3);
Console.WriteLine("После удаления третьей строки:");
Console.WriteLine("Сумма всех строк: " + allLines.GetTotalSum());

public interface IDeliverable
{
    string Address { get; set; }
    void Deliver();
}

public interface IReturnable
{
    int ReturnDays { get; set; }
    void ReturnItem();
}

public class OrderLine
{
    private int productId;
    private string productName;
    private decimal price;

    public int ProductId
    {
        get { return productId; }
        set { productId = value; }
    }

    public string ProductName
    {
        get { return productName; }
        set { productName = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public string Category { get; set; }
    public bool IsPaid { get; set; }
    public string Comment { get; set; }

    public OrderLine(int productId, string productName, decimal price)
    {
        ProductId = productId;
        ProductName = productName;
        Price = price;
        Category = "Общая";
        IsPaid = false;
        Comment = "Нет";
    }

    public virtual decimal CalculateTotal()
    {
        return Price;
    }

    public virtual void UpdatePrice(decimal newPrice)
    {
        Price = newPrice;
    }

    public void UpdatePrice(decimal newPrice, string reason)
    {
        UpdatePrice(newPrice);
        Comment = reason;
        Console.WriteLine("Причина изменения цены: " + reason);
    }

    public virtual void GetProductDetails()
    {
        Console.WriteLine("ID товара: " + ProductId);
        Console.WriteLine("Название товара: " + ProductName);
        Console.WriteLine("Цена товара: " + Price);
    }

    public void MarkAsPaid()
    {
        IsPaid = true;
        Console.WriteLine("Строка оплачена: " + ProductName);
    }

    public bool IsExpensive()
    {
        return Price > 1000;
    }

    public void ShowStatus()
    {
        Console.WriteLine("Категория: " + Category);
        Console.WriteLine("Оплачено: " + IsPaid);
        Console.WriteLine("Комментарий: " + Comment);
    }

    public void ShowStatus(string title)
    {
        Console.WriteLine(title);
        ShowStatus();
    }
}

public class StandardLine : OrderLine
{
    private int units;

    public int Units
    {
        get { return units; }
        set { units = value; }
    }

    public int Weight { get; set; }
    public string Size { get; set; }
    public string Packaging { get; set; }

    public StandardLine(int productId, string productName, decimal price, int units) : base(productId, productName, price)
    {
        Units = units;
        Weight = 1;
        Size = "Средний";
        Packaging = "Коробка";
    }

    public override decimal CalculateTotal()
    {
        return Price * Units;
    }

    public decimal CalculateTotal(int extraUnits)
    {
        return Price * (Units + extraUnits);
    }

    public override void GetProductDetails()
    {
        base.GetProductDetails();
        Console.WriteLine("Количество единиц: " + Units);
    }

    public int GetTotalWeight()
    {
        return Weight * Units;
    }

    public bool IsBulk()
    {
        return Units >= 10;
    }

    public void ShowPackaging()
    {
        Console.WriteLine("Размер: " + Size);
        Console.WriteLine("Упаковка: " + Packaging);
    }
}

public class SpecialLine : OrderLine
{
    private decimal discount;

    public decimal Discount
    {
        get { return discount; }
        set { discount = value; }
    }

    public string PromoCode { get; set; }
    public int PromoDays { get; set; }
    public bool IsActive { get; set; }

    public SpecialLine(int productId, string productName, decimal price, decimal discount) : base(productId, productName, price)
    {
        Discount = discount;
        PromoCode = "Нет";
        PromoDays = 7;
        IsActive = true;
    }

    public override void UpdatePrice(decimal newPrice)
    {
        Price = newPrice - (newPrice * Discount / 100);
    }

    public override void GetProductDetails()
    {
        base.GetProductDetails();
        Console.WriteLine("Скидка: " + Discount + "%");
    }

    public void ShowPromo()
    {
        Console.WriteLine("Промокод: " + PromoCode);
        Console.WriteLine("Дней действия: " + PromoDays);
        Console.WriteLine("Акция активна: " + IsActive);
    }

    public bool IsBigDiscount()
    {
        return Discount >= 20;
    }

    public decimal GetDiscountAmount()
    {
        return Price * Discount / 100;
    }

    public decimal GetDiscountAmount(int units)
    {
        return Price * units * Discount / 100;
    }
}

public class FreeLine : OrderLine
{
    private decimal prepayment;

    public decimal Prepayment
    {
        get { return prepayment; }
        set { prepayment = value; }
    }

    public string Reason { get; set; }
    public string Giver { get; set; }
    public int Priority { get; set; }

    public FreeLine(int productId, string productName, decimal price, decimal prepayment) : base(productId, productName, price)
    {
        Prepayment = prepayment;
        Reason = "Нет";
        Giver = "Магазин";
        Priority = 1;
    }

    public override decimal CalculateTotal()
    {
        return Price - Prepayment;
    }

    public override void GetProductDetails()
    {
        base.GetProductDetails();
        Console.WriteLine("Предварительный платеж: " + Prepayment);
    }

    public bool IsFullyPrepaid()
    {
        return Prepayment >= Price;
    }

    public decimal GetDebt()
    {
        return Price - Prepayment;
    }

    public decimal GetDebt(decimal extraPayment)
    {
        return Price - Prepayment - extraPayment;
    }

    public void ShowReason()
    {
        Console.WriteLine("Причина: " + Reason);
        Console.WriteLine("Кто выдал: " + Giver);
        Console.WriteLine("Приоритет: " + Priority);
    }
}

public class WholesaleLine : StandardLine
{
    public int WholesaleDiscount { get; set; }
    public string Warehouse { get; set; }
    public int MinUnits { get; set; }

    public WholesaleLine(int productId, string productName, decimal price, int units, int wholesaleDiscount)
        : base(productId, productName, price, units)
    {
        WholesaleDiscount = wholesaleDiscount;
        Warehouse = "Основной";
        MinUnits = 10;
    }

    public override decimal CalculateTotal()
    {
        decimal total = base.CalculateTotal();
        return total - (total * WholesaleDiscount / 100);
    }

    public override void GetProductDetails()
    {
        base.GetProductDetails();
        Console.WriteLine("Оптовая скидка: " + WholesaleDiscount + "%");
        Console.WriteLine("Склад: " + Warehouse);
    }

    public decimal GetSavedMoney()
    {
        return base.CalculateTotal() - CalculateTotal();
    }

    public bool IsWholesale()
    {
        return Units >= MinUnits;
    }

    public void ShowWarehouse()
    {
        Console.WriteLine("Склад: " + Warehouse);
        Console.WriteLine("Минимум единиц для опта: " + MinUnits);
    }
}

public class DeliveryLine : OrderLine, IDeliverable, IReturnable
{
    public string Address { get; set; }
    public int ReturnDays { get; set; }
    public decimal DeliveryCost { get; set; }
    public string Courier { get; set; }

    public DeliveryLine(int productId, string productName, decimal price, string address, int returnDays)
        : base(productId, productName, price)
    {
        Address = address;
        ReturnDays = returnDays;
        DeliveryCost = 0;
        Courier = "Не назначен";
    }

    public void Deliver()
    {
        Console.WriteLine("Доставка: " + ProductName + ", адрес: " + Address);
    }

    public void Deliver(string newAddress)
    {
        Address = newAddress;
        Deliver();
    }

    public void ReturnItem()
    {
        Console.WriteLine("Возврат: " + ProductName + ", дней на возврат: " + ReturnDays);
    }

    public decimal GetTotalWithDelivery()
    {
        return CalculateTotal() + DeliveryCost;
    }

    public decimal GetTotalWithDelivery(decimal extraCost)
    {
        return CalculateTotal() + DeliveryCost + extraCost;
    }

    public override void GetProductDetails()
    {
        base.GetProductDetails();
        Console.WriteLine("Адрес доставки: " + Address);
        Console.WriteLine("Дней на возврат: " + ReturnDays);
        Console.WriteLine("Курьер: " + Courier);
    }
}

public class OrderCollection<T> where T : OrderLine
{
    private List<T> lines = new List<T>();

    public void Add(T line)
    {
        lines.Add(line);
    }

    public void Remove(T line)
    {
        lines.Remove(line);
    }

    public void DisplayLines()
    {
        foreach (var line in lines)
        {
            line.GetProductDetails();
            Console.WriteLine("Итого по строке: " + line.CalculateTotal());
        }
    }

    public decimal GetTotalSum()
    {
        decimal total = 0;
        foreach (var line in lines)
        {
            total = total + line.CalculateTotal();
        }
        return total;
    }
}
```
