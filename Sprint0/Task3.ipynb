Создать базовый класс Order в C#, который будет представлять информацию о
заказах товаров или услуг. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма. 


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Item
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Item(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
}

public class Order
{
    // Атрибуты
    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public decimal TotalAmount { get; protected set; }

    // Список элементов заказа
    public List<Item> Items { get; protected set; } = new List<Item>();

    public Order(int orderId)
    {
        OrderId = orderId;
        CreationDate = DateTime.Now;
        TotalAmount = 0;
    }

    public virtual decimal CalculateTotal()
    {
        decimal total = 0;
        foreach (var item in Items)
        {
            total += item.Price;
        }
        TotalAmount = total;
        return TotalAmount;
    }

    public virtual void AddItem(Item item)
    {
        Items.Add(item);
        CalculateTotal();
        Console.WriteLine($"Добавлен товар: {item.Name}");
    }

    public virtual void RemoveItem(Item item)
    {
        if (Items.Remove(item))
        {
            CalculateTotal();
            Console.WriteLine($"Удален товар: {item.Name}");
        }
    }
}

public class OnlineOrder : Order
{
    public string CustomerEmail { get; set; }
    public string DeliveryMethod { get; set; }

    public OnlineOrder(int orderId, string customerEmail, string deliveryMethod) 
        : base(orderId)
    {
        CustomerEmail = customerEmail;
        DeliveryMethod = deliveryMethod;
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"   (Онлайн-заказ) Доставка способом: {DeliveryMethod}. Клиент: {CustomerEmail}");
    }
}


public class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }

    public PhysicalOrder(int orderId, string deliveryAddress) 
        : base(orderId)
    {
        DeliveryAddress = deliveryAddress;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"   [Физический заказ] Товар возвращен по адресу: {DeliveryAddress}");
    }
}

public class SpecializedOrder : Order
{
    public string SpecialConditions { get; set; }
    public decimal DiscountAmount { get; set; }

    public SpecializedOrder(int orderId, string specialConditions, decimal discountAmount) 
        : base(orderId)
    {
        SpecialConditions = specialConditions;
        DiscountAmount = discountAmount;
    }

    public override decimal CalculateTotal()
    {
        decimal baseTotal = base.CalculateTotal();
        TotalAmount = baseTotal - DiscountAmount;
        if (TotalAmount < 0) TotalAmount = 0;
        return TotalAmount;
    }
}

class Program
{
    static void Main()
    {
        Item item1 = new Item("Клавиатура", 3000);
        Item item2 = new Item("Мышь", 1500);

        // 1. Онлайн-заказ
        OnlineOrder onlineOrder = new OnlineOrder(1, "1test@gmail.com", "Курьер");
        onlineOrder.AddItem(item1);

        // 2. Физический заказ
        PhysicalOrder physicalOrder = new PhysicalOrder(2, "ул. Пушкина, д. 10");
        physicalOrder.AddItem(item1);
        physicalOrder.RemoveItem(item1);

        // 3. Специализированный заказ
        SpecializedOrder specOrder = new SpecializedOrder(3, "Скидка постоянного клиента", 500);
        specOrder.AddItem(item1);
        specOrder.AddItem(item2);
        Console.WriteLine($"Сумма со специальным условием: {specOrder.TotalAmount}");
    }
}
```
