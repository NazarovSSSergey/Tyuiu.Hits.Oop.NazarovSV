<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс OrderLine в C#, который будет представлять информацию о строке заказа, содержащей детали одного товара в заказе. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
1. СтандартнаяСтрока (StandardLine): Должна содержать дополнительные атрибуты, такие как Количество единиц (Units). Метод CalculateTotal() должен быть переопределен для учета количества единиц при расчете общей стоимости.
2. СпециальнаяСтрока (SpecialLine): Должна содержать дополнительные атрибуты, такие как Скидка (Discount). Метод UpdatePrice() должен быть переопределен для применения скидки к цене товара.
3. БесплатнаяСтрока (FreeLine) (если требуется третий класс): Должна содержать дополнительные атрибуты, такие как Предварительный платеж (Prepayment). Метод CalculateTotal() должен быть переопределен для учета предварительного плата при расчете общей стоимости.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями 


<h2 style="color:DodgerBlue">Реализация:</h2>

```
OrderLine line = new OrderLine { ProductId = 1, ProductName = "Ручка", Price = 50, Category = "Канцелярия", Article = "A-001", IsAvailable = true };
OrderLine standard = new StandardLine { ProductId = 2, ProductName = "Тетрадь", Price = 500, Category = "Канцелярия", Article = "A-002", IsAvailable = true, Units = 3, Warehouse = "Центральный", DeliveryDays = 2, Weight = 0.4m };
OrderLine special = new SpecialLine { ProductId = 3, ProductName = "Рюкзак", Price = 1000, Category = "Сумки", Article = "B-003", IsAvailable = true, Discount = 10, PromoCode = "SALE10", ValidDays = 7, Reason = "Акция" };
OrderLine free = new FreeLine { ProductId = 4, ProductName = "Калькулятор", Price = 800, Category = "Электроника", Article = "C-004", IsAvailable = true, Prepayment = 200, GiftName = "Чехол", GiftPrice = 150 };

StandardLine standardLine = (StandardLine)standard;
SpecialLine specialLine = (SpecialLine)special;
FreeLine freeLine = (FreeLine)free;

OrderService lineService = new OrderService(line);
OrderService standardService = new OrderService(standard);
OrderService specialService = new OrderService(special);
OrderService freeService = new OrderService(free);

Console.WriteLine("--- Исходные строки заказа ---");
lineService.ShowOrder();
standardService.ShowOrder();
specialService.ShowOrder();
freeService.ShowOrder();

Console.WriteLine("--- Новая цена товара: 60, 550, 1200, 900 ---");
line.UpdatePrice(60);
standard.UpdatePrice(550);
special.UpdatePrice(1200);
free.UpdatePrice(900);
Console.WriteLine("СпециальнаяСтрока: цена товара 1200, скидка " + specialLine.Discount + "%, цена товара после скидки: " + specialLine.Price);
lineService.ShowOrder();
standardService.ShowOrder();
specialService.ShowOrder();
freeService.ShowOrder();

Console.WriteLine("--- OrderLine ---");
line.SetAvailability(false);
Console.WriteLine(line.GetCategoryInfo() + ", дорогой: " + (line.IsExpensive(100) ? "да" : "нет") + ", в наличии: " + (line.IsAvailable ? "да" : "нет"));

Console.WriteLine("--- СтандартнаяСтрока ---");
Console.WriteLine("До: количество единиц " + standardLine.Units + ", общая стоимость " + standardLine.CalculateTotal());
standardLine.AddUnits(2);
Console.WriteLine("После: количество единиц " + standardLine.Units + ", общая стоимость " + standardLine.CalculateTotal());
Console.WriteLine(standardLine.GetDeliveryInfo() + ", цена товара за единицу: " + standardLine.GetPricePerUnit());

Console.WriteLine("--- СпециальнаяСтрока: скидка 15% ---");
((IDiscountable)special).ApplyDiscount(15);
specialLine.UpdatePrice(1200);
Console.WriteLine("СпециальнаяСтрока: цена товара 1200, скидка " + specialLine.Discount + "%, цена товара после скидки: " + specialLine.Price);
Console.WriteLine(specialLine.GetPromoInfo() + ", экономия: " + specialLine.GetSavings());
specialService.ShowOrder();

Console.WriteLine("--- БесплатнаяСтрока ---");
Console.WriteLine(freeLine.GetGiftInfo() + ", предварительный платеж внесен: " + (freeLine.IsPrepaid() ? "да" : "нет") + ", остаток: " + freeLine.GetRemainingPayment());
freeService.ShowOrder();

public interface IOrderLine
{
    decimal CalculateTotal();
    string GetProductDetails();
    void PrintLine();
}

public interface IDiscountable
{
    void ApplyDiscount(decimal percent);
}

public class OrderLine : IOrderLine
{
    public int ProductId { get; set; }
    public string ProductName { get; set; }
    public decimal Price { get; set; }
    public string Category { get; set; }
    public string Article { get; set; }
    public bool IsAvailable { get; set; }

    public virtual decimal CalculateTotal()
    {
        return Price;
    }

    public virtual void UpdatePrice(decimal newPrice)
    {
        Price = newPrice;
    }

    public virtual string GetProductDetails()
    {
        return "ID товара: " + ProductId + ", название товара: " + ProductName + ", цена товара: " + Price + ", категория: " + Category + ", артикул: " + Article;
    }

    public void SetAvailability(bool available)
    {
        IsAvailable = available;
    }

    public string GetCategoryInfo()
    {
        return "Категория: " + Category;
    }

    public bool IsExpensive(decimal limit)
    {
        return Price > limit;
    }

    void IOrderLine.PrintLine()
    {
        Console.WriteLine(GetProductDetails() + " | общая стоимость: " + CalculateTotal());
    }
}

public class StandardLine : OrderLine
{
    public int Units { get; set; }
    public string Warehouse { get; set; }
    public int DeliveryDays { get; set; }
    public decimal Weight { get; set; }

    public override decimal CalculateTotal()
    {
        return Price * Units;
    }

    public override string GetProductDetails()
    {
        return base.GetProductDetails() + ", количество единиц: " + Units;
    }

    public void AddUnits(int count)
    {
        Units += count;
    }

    public string GetDeliveryInfo()
    {
        return "Склад " + Warehouse + ", доставка " + DeliveryDays + " дн., вес: " + Weight + " кг";
    }

    public decimal GetPricePerUnit()
    {
        return Price;
    }
}

public class SpecialLine : OrderLine, IDiscountable
{
    public decimal Discount { get; set; }
    public string PromoCode { get; set; }
    public int ValidDays { get; set; }
    public string Reason { get; set; }

    public override void UpdatePrice(decimal newPrice)
    {
        Price = newPrice - newPrice * Discount / 100;
    }

    public override string GetProductDetails()
    {
        return base.GetProductDetails() + ", скидка: " + Discount + "%";
    }

    public string GetPromoInfo()
    {
        return "Промокод " + PromoCode + " на " + ValidDays + " дн., причина скидки: " + Reason;
    }

    public decimal GetSavings()
    {
        return Price * Discount / 100;
    }

    void IDiscountable.ApplyDiscount(decimal percent)
    {
        Discount = percent;
    }
}

public class FreeLine : OrderLine
{
    public decimal Prepayment { get; set; }
    public string GiftName { get; set; }
    public decimal GiftPrice { get; set; }

    public override decimal CalculateTotal()
    {
        return base.CalculateTotal() - Prepayment;
    }

    public override string GetProductDetails()
    {
        return base.GetProductDetails() + ", предварительный платеж: " + Prepayment;
    }

    public bool IsPrepaid()
    {
        return Prepayment > 0;
    }

    public string GetGiftInfo()
    {
        return "Подарок: " + GiftName + ", стоимость: " + GiftPrice;
    }

    public decimal GetRemainingPayment()
    {
        return Price - Prepayment;
    }
}

public class OrderService
{
    private readonly IOrderLine orderLine;

    public OrderService(IOrderLine line)
    {
        orderLine = line;
    }

    public void ShowOrder()
    {
        orderLine.PrintLine();
    }
}
```
