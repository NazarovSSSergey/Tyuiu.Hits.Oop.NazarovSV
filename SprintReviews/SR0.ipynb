```
using System;

Plant plant1 = new Plant("Трава", "Растение", 10.5);
Flower flower1 = new Flower("Роза", "Цветок", 30.5, "Красный", 24);
Tree tree1 = new Tree("Дуб", "Дерево", 500.0, 15, "Лиственные");

plant1.GetPlantDetails();
plant1.Grow(2.5);

flower1.GetPlantDetails();
flower1.Smell();
flower1.Grow(5.5);

tree1.GetPlantDetails();
tree1.Grow(20.0);
tree1.GrowUp(5);

Plant[] garden = { plant1, flower1, tree1 };
foreach (Plant plant in garden)
{
    plant.Bloom();
}

public class Plant
{
    public string Name;
    public string Type;
    public double Height;

    public Plant(string name, string type, double height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine("Растение " + Name + " цветет");
    }

    public virtual void Grow(double growth)
    {
        Height += growth;
        Console.WriteLine(Name + " вырос на " + growth + " см. Высота: " + Height + " см");
    }

    public virtual void GetPlantDetails()
    {
        Console.WriteLine(Name + " | Тип: " + Type + " | Высота: " + Height + " см");
    }
}

public class Flower : Plant
{
    public string Color;
    public int PetalCount;

    public Flower(string name, string type, double height, string color, int petalCount) : base(name, type, height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine("Цветок " + Name + " распускается и радует глаз!");
    }

    public void Smell()
    {
        Console.WriteLine("Цветок " + Name + " начинает пахнуть");
    }

    public void Wither()
    {
        Console.WriteLine("Цветок " + Name + " начинает увядать");
    }

    public override void GetPlantDetails()
    {
        base.GetPlantDetails();
        Console.WriteLine("Цвет: " + Color + " | Лепестков: " + PetalCount);
    }
}

public class Tree : Plant
{
    public int Age;
    public string LeafType;

    public Tree(string name, string type, double height, int age, string leafType) : base(name, type, height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine("Дерево " + Name + " покрывается новыми листьями!");
    }

    public void GrowUp(int years)
    {
        Age += years;
        Console.WriteLine(Name + " стал старше на " + years + " лет. Возраст: " + Age + " лет");
    }

    public override void GetPlantDetails()
    {
        base.GetPlantDetails();
        Console.WriteLine("Возраст: " + Age + " лет | Листья: " + LeafType);
    }
}
```
